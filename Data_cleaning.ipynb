```python
missing_values = df.isnull().sum()

print(missing_values)
```

```python
# Duplicate Rows
duplicate_rows = df.duplicated().sum()

print("Number of Duplicate Rows:", duplicate_rows)
```

```python
# Unique Values
for column in df.columns:
    print(
        column,
        "→",
        df[column].nunique(),
        "unique values"
    )
```

```python
df.dtypes
```

```python
df["date"] = pd.to_datetime(df["date"], errors="coerce")
```

```python
print(df["date"].dtype)
```

```python
# Date Range 
print("Starting Date: ",df["date"].min())
print("Ending Date: ",df["date"].max())
```

```python
invalid_dates = df["date"].isnull().sum()

print("Invalid Dates:", invalid_dates)
```

```python
# numericals columns
numeric_columns = df.select_dtypes(
    include=np.number
).columns

print("Numerical Columns:")
print(numeric_columns.tolist())
```

```python
# Check Negative Values
for column in numeric_columns:
    negative_values = (df[column] < 0).sum()

    print(
        column,
        "→",
        negative_values,
        "negative values"
    )
```

```python
median_demand = df["demand_forecast"].median()

print("Median Demand Forecast:", median_demand)
```

```python
df["demand_forecast"] = df["demand_forecast"].apply(
    lambda x: median_demand if x < 0 else x
)
```

```python
print(
    "Negative Demand Forecast values:",
    (df["demand_forecast"] < 0).sum()
)
```

```python
negative_count = (df["demand_forecast"] < 0).sum()

print("Negative values replaced:", negative_count)
print("Replacement value (Median):", median_demand)
```

```python
categorical_columns = df.select_dtypes(
    include="object"
).columns

for column in categorical_columns:
    print("\n", column)
    print(df[column].unique()[:20])
```
